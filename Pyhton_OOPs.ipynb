```python
class PersonDetails:
    def printHello(self):
        print("Hello");

    def __init__(self, name):
        self.name = name

    def printPersonNameFromSelf(self):
        print(f"Person name is {self.name}")

    @classmethod
    def printPersonNameFromClassMethod(cls, name):
        print(f"Person name is {name}")

    @staticmethod
    def printOfficeAddressForPerson(name):
        print(f"{name} address")
```

```python
# Creating an object
p = PersonDetails("Mahima")
```

```python
p.printPersonNameFromSelf()
p.printPersonNameFromClassMethod('Vallabh')
```

```python
PersonDetails.printOfficeAddressForPerson("Mahima")
```

```python
#Inheritence
class Employee:
  def __init__(self, name, address):
      self.name = name
      self.address = address

  def display_details(self):
    print(f"Name of the Employee {self.name} & Address {self.address}")
```

```python
class Manager(Employee):
  def __init__(self, name, address, teamSize):
     super().__init__(name, address)
     self.teamSize = teamSize

  def display_details(self):
    super().display_details()
    print(f"Manger team Size {self.teamSize}")
```

```python
m1 = Manager("Mahima", "Bangalore", 5)
m1.display_details()
#m1.display_manager_details()
```

```python
#Polymorphism
class Developer:
    def role(self):
        print("Writes code")

class Designer:
    def role(self):
        print("Designs UI")

# Polymorphic behaviour
def show_role(person):
    person.role()

d1 = Developer()
d2 = Designer()

show_role(d1)  # Output: Writes code
show_role(d2)  # Output: Designs UI
```

```python
#Polymorphism with Inheritance
class Animal:
    def speak(self):
        print("Some sound")

class Dog(Animal):
    def speak(self):
        print("Bark")

class Cat(Animal):
    def speak(self):
        print("Meow")

animals = [Dog(), Cat()]

for animal in animals:
    animal.speak()
```

```python
#Python supports abstraction primarily through abstract base classes (ABCs) and interfaces, using the abc module.
from abc import ABC, abstractmethod

class TeamMember(ABC):

    @abstractmethod
    def role(self):
        pass
```

```python
#The TeamMember class defines an abstract method role(), which means that any class inheriting from TeamMember must implement this method.
class Developer(TeamMember):
    def role(self):
        print("Writes code")

class Designer(TeamMember):
    def role(self):
        print("Designs UI")
```

```python
#Now we can use these classes in a unified way:
members = [Developer(), Designer()]

for member in members:
    member.role()
```
