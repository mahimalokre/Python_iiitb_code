```python
# 1. Homogeneous vs. Heterogeneous Data Types
# Python List: A Python list can contain elements of different types.
my_list = [1, 2.5, "hello", True]
print(my_list)  # [1, 2.5, 'hello', True]

# NumPy Array: NumPy tries to convert all elements to a single data type (here, string), so the array is homogeneous.
import numpy as np

my_array = np.array([1, 2.5, "hello", True])
print(my_array)  # ['1' '2.5' 'hello' 'True']
print(my_array.dtype)  # <U32 (string type)
```

```python
# 2. Arrays of Varying Sizes
# Python List: You can have lists of different lengths inside a list.

list_of_lists = [[1, 2, 3], [4, 5], [6]]
print(list_of_lists)  # [[1, 2, 3], [4, 5], [6]]

# NumPy Array: NumPy cannot create a true multidimensional array with varying lengths; it falls back to dtype object,
# losing the benefits of fast computation.

array_of_arrays = np.array([[1, 2, 3], [4, 5], [6]])
print(array_of_arrays)
# Output: array([list([1, 2, 3]), list([4, 5]), list([6])], dtype=object)
```

```python
# 3. Optimized Mathematical Operations
# Python List:


list1 = [1, 2, 3]
list2 = [4, 5, 6]
# You can't do list1 + list2 for addition; it concatenates
print(list1 + list2)  # [1, 2, 3, 4, 5, 6]
# Element-wise addition requires a loop or comprehension
result = [x + y for x, y in zip(list1, list2)]
print(result)  # [5, 7, 9]

# NumPy Array: In NumPy, mathematical operations like addition, subtraction, multiplication, etc.
# ,are broadcasted element-wise without the need for loops.


array1 = np.array([1, 2, 3])
array2 = np.array([4, 5, 6])
# Element-wise addition
result = array1 + array2
print(result)  # [5 7 9]
```

```python
# 4. Broadcasting: Broadcasting allows NumPy arrays of different shapes to be used together for arithmetic operations.
# Python List:

list1 = [1, 2, 3]
scalar = 2
# You cannot multiply a scalar directly element-wise; you need a loop or comprehension
result = [x * scalar for x in list1]
print(result)  # [2, 4, 6]

# NumPy Array: NumPy arrays are optimized for broadcasting, enabling operations between arrays and scalars or arrays of different shapes without explicit loops.

array = np.array([1, 2, 3])
scalar = 2
# Scalar multiplication is directly broadcasted
result = array * scalar
print(result)  # [2 4 6]
```

```python
# 5. Matrix Multiplication
# Python List:


list1 = [[1, 2], [3, 4]]
list2 = [[5, 6], [7, 8]]
# You need nested loops for matrix multiplication
result = [[sum(a * b for a, b in zip(row, col)) for col in zip(*list2)] for row in list1]
print(result)  # [[19, 22], [43, 50]]

# NumPy Array: NumPy makes matrix multiplication simple and efficient using the @ operator or np.dot.


array1 = np.array([[1, 2], [3, 4]])
array2 = np.array([[5, 6], [7, 8]])
# Matrix multiplication is straightforward using @ or np.dot
result = array1 @ array2  # Alternatively, np.dot(array1, array2)
print(result)  # [[19 22]
               #  [43 50]]
```
